```python
import numpy as np
import pandas as pd
```

```python
df1=pd.DataFrame({
    'A':['A0','A1','A2'],
    'B':['B0','B1','B2'],
    'C':['C0','C1','C2']
})
df2=pd.DataFrame({
    'A':['A3','A4','A5'],
    'B':['B3','B4','B5'],
    'C':['C3','C4','C5']
})
```

```python
df1
```

```python
df2
```

```python
#concat through column(by default axis=0)
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2],axis=1)
```

```python
#join two data frame
df3=pd.DataFrame({
    'name':['Alice','Bob','Charlic']
},index=[1,2,3])
df4=pd.DataFrame({
    'score':[85,90,75]
},index=[2,3,4])
```

```python
df3
```

```python
df4
```

```python
df3.join(df4)
```

```python
df3.join(df4,how='outer')
```

```python
df4.join(df3)
```
